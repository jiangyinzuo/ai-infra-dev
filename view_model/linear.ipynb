# nn.Linear

全连接层 / 仿射变换，即$$y = xW^T + b$$，向量x矩阵。把一个n维特征向量输入，通过向量(1, n)乘矩阵(n, m)，变成一个m维特征向量输出。

`W` 做的事情可以理解为：

- 输入向量 `x` 的每一个维度，是一个"特征"（比如像素亮度、词向量某一维、传感器读数）。
- 输出向量 `y` 的每一个维度，是输入特征的一种加权组合。

例如：

$$
y_1 = 0.2·x_1 + 0.8·x_2 - 0.1·x_3 \\
y_2 = -0.5·x_1 + 0.3·x_2 + 0.9·x_3
$$

batch计算：输入张量`x`的前面n-1维度都可以统一视为$$dim_1 * dim_2 * ... * dim_{n-1}$$个向量，批量做向量x矩阵，即矩阵x矩阵。

```python
import torch
import torch.nn as nn
x = torch.randn(32, 128)        # batch_size=32, in_features=128
linear = nn.Linear(128, 64)
y = linear(x)
y.shape
```
